```python
from time import sleep
import gymnasium as gym
import random
env = gym.make('MountainCar-v0', render_mode='human')

seed = 0
random.seed(seed)

print(f'Action Space: {env.action_space}')
print(f'Observation Space: {env.observation_space}')

episodes=200
state,_ = env.reset()  # Reset the environment
reward_sum=0

for episode in range(episodes):
    env.render()  # Visualize the environment

    position, velocity = state  # Unpack state (position, velocity)

    # Improved Policy: Build momentum to the right to reach the top
    if velocity < 0:
        action = 0  # Push left if velocity is negative
    elif velocity >= 0:
        if position < 0:  # Car is in the valley, needs to build momentum
            action = 2  # Push right to gain momentum
        else:  # Car is getting closer to the goal
            action = 0  # No action or apply a small force

    # Taking the action and getting the new state
    state, reward, done, truncated, info = env.step(action)
    reward_sum += reward

    # Environment information
    print(f"Position: {position:.3f}, Velocity: {velocity:.3f}, Action: {action}, Reward: {reward}")
    print(f"Reward sum: {reward_sum}")

    # Break if the episode is done
    if done or truncated:
        print(f"Episode ended with total reward: {reward_sum}")
        break

    sleep(0.01)

env.close()
```

```python
env = gym.make("ALE/Breakout-v5", render_mode="rgb_array")
obs, info = env.reset()

raw_frame = env.render()
print(type(raw_frame))
print(raw_frame.shape)
```

```python
# convert to RGB scale 
import cv2

def to_grayScale(frame_rgb):
    return cv2.cvtColor(frame_rgb, cv2.COLOR_RGB2GRAY)

gray = to_grayScale(raw_frame)
print(gray.shape)
```

```python
#Step4 Down-size sample to 110,84

def downSample(frame_gray):
    return cv2.resize(frame_gray, (84,110), interpolation =cv2.INTER_LINEAR) 
    # resize takes width, height. 
```

```python
downsized = downSample(gray)
print(downsized.shape)
```

```python
# crop the image to 84 x84 

def crop(frame_110x84):
    top = 110-84
    return frame_110x84[top:top + 84, 0:84]  # ndarray operations 
```

```python
cropped = crop(downsized)
print(cropped.shape)
```

```python
# stacking the last four frames  
   # this is required because single frame will not tell the direction and velocity of the ball or enemy velocity.  stacking last 4 will give implict
    # motion positon without needeing recurrent architecture 
from collections import deque
import numpy as np

frame_stack = deque(maxlen=4)

def stackFrames(new_cropped_frame):
    frame_stack.append(new_cropped_frame)
    while len(frame_stack) < 4 :
        frame_stack.append(new_cropped_frame)
    return np.stack(frame_stack, axis=-1)

state = stackFrames(cropped)
print(state.shape)
print(type(state))
```
